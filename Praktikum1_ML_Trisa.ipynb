```python
from google.colab import drive
drive.mount('/content/drive')

import pandas as pd

df = pd.read_csv('/content/drive/MyDrive/ML 2026/data/day.csv')

df.head()
```

```python
# Dimension Date
dim_date = df[
    ['instant', 'dteday', 'yr', 'mnth', 'season',
     'weekday', 'holiday', 'workingday']
].copy()

dim_date = dim_date.rename(columns={
    'instant': 'date_id',
    'dteday': 'date'
})

print("DIMENSION DATE")
display(dim_date.head())
```

```python
# Dimension Weather

dim_weather = df[
    ['weathersit']
].drop_duplicates().sort_values('weathersit').copy()

dim_weather['weather_id'] = range(1, len(dim_weather) + 1)

dim_weather = dim_weather[
    ['weather_id', 'weathersit']
]

print("DIMENSION WEATHER")
display(dim_weather)
```

```python
# Fact Bike Rental

fact_bike_rental = df[
    ['instant', 'weathersit', 'temp', 'atemp',
     'hum', 'windspeed', 'casual',
     'registered', 'cnt']
].copy()

fact_bike_rental = fact_bike_rental.rename(columns={
    'instant': 'date_id'
})

fact_bike_rental['weather_id'] = fact_bike_rental['weathersit'].map(
    dim_weather.set_index('weathersit')['weather_id']
)

fact_bike_rental = fact_bike_rental.drop(columns=['weathersit'])

fact_bike_rental = fact_bike_rental[
    ['date_id', 'weather_id', 'temp', 'atemp',
     'hum', 'windspeed', 'casual',
     'registered', 'cnt']
]

print("FACT BIKE RENTAL")
display(fact_bike_rental.head())
```

```python
# Pivot Table: Total Penyewaan Berdasarkan Bulan dan Tahun

pivot_month = pd.pivot_table(
    df,
    values='cnt',
    index='mnth',
    columns='yr',
    aggfunc='sum'
)

print("PIVOT TABLE - TOTAL PENYEWAAN BERDASARKAN BULAN DAN TAHUN")
display(pivot_month)
```

```python
import matplotlib.pyplot as plt

pivot_month.plot(
    kind='bar',
    figsize=(10, 5),
    title='Total Penyewaan Sepeda Berdasarkan Bulan dan Tahun',
    xlabel='Bulan',
    ylabel='Total Penyewaan'
)

plt.show()
```
